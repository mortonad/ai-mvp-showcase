```python
import os
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY') if 'google.colab' in str(get_ipython()) else "YOUR_API_KEY"

llm = ChatOpenAI(model="gpt-4-turbo-preview", temperature=0.7)

def bot_process_message(user_name, message):
    system_prompt = """
    你是一位 AI 行銷專家，說話風格親切、有溫度，略帶一點點幽默。
    你的目標是讓用戶感受到 Morton 的專業與誠意，並促成進一步的合作諮詢。
    """
    messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"用戶姓名：{user_name}\n訊息內容：{message}")
    ]
    response = llm.invoke(messages)
    return response.content

print("行銷 Bot 實戰邏輯載入完成")
```

```python
# 實測三種不同情境
test_cases = [
    ("王小姐", "我想詢問線上課程的價格？"),
    ("李先生", "我是室內設計師，想了解怎麼用 AI 幫我找案子？"),
    ("陳老闆", "你們有在接企業內訓嗎？")
]

for name, msg in test_cases:
    print(f"--- 測試情境：{name} ---")
    print(f"用戶訊息：{msg}")
    print(f"AI 自動回覆：\n{bot_process_message(name, msg)}")
    print("-" * 40)
```
